## 9. Build a Predictive Model

So far we have described the penguins we already measured.

Now we will build a **machine learning model** that tries to predict something we do *not* already know: given a penguin's measurements, which **species** is it?

The idea is simple:

1. Show the model a set of penguins where we already know the species (the **training** data).
2. Hide a second set of penguins from it (the **testing** data).
3. Ask the model to predict the species of the hidden penguins.
4. Compare its predictions to the real answers and score how well it did.

We are previewing an idea here that we will study in depth later in the semester. For today, focus on *reading* the code and being able to say what each line is doing.

### 9.1 Import the Machine Learning Tools

`scikit-learn` (imported as `sklearn`) is Python's standard machine learning library. It is already installed in Colab.

````python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import f1_score
from sklearn.metrics import roc_auc_score
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix
from sklearn.metrics import ConfusionMatrixDisplay

print("Machine learning tools imported!")
````

### 9.2 Handle the Missing Values

Look back at row 3 of the dataset. Several columns contain `NaN`, which means the value is **missing**.

A model cannot train on missing values, so we need to deal with them. The simplest option is to drop any row that has a missing value.

`dropna()` returns a new DataFrame with those rows removed.

````python
model_df = df.dropna()

print("Rows before dropping missing values:", len(df))
print("Rows after dropping missing values: ", len(model_df))
print("Rows removed:", len(df) - len(model_df))
````

*(ANSWER — expected output)*

```
Rows before dropping missing values: 344
Rows after dropping missing values:  333

```

Eleven rows go away: 2 penguins missing every measurement, and 9 more missing only `sex`. Dropping is the quick answer for Week 1 — later in the course we will talk about *imputing* missing values instead, and about why dropping rows can quietly bias a dataset.

### 9.3 Choose the Target and the Features

Two terms we will use all semester:

- The **target** (`y`) is the thing we are trying to predict. Here it is `species`.
- The **features** (`X`) are the pieces of information the model gets to look at.

One complication: a model does math, so it cannot work with text like `"Biscoe"` or `"MALE"` directly. `pd.get_dummies()` converts each text category into its own True/False column. This is called **one-hot encoding**.

````python
target_column = "species"

feature_columns = [
    "island",
    "bill_length_mm",
    "bill_depth_mm",
    "flipper_length_mm",
    "body_mass_g",
    "sex"
]

y = model_df[target_column]
X = model_df[feature_columns]

# Turn the text columns into numeric True/False columns
X = pd.get_dummies(X, columns=["island", "sex"])

X.head()
````

Notice that `island` and `sex` are gone, replaced by a set of new columns.

Note that `species` is **not** in `X`. If we gave the model the answer as a feature, it would score perfectly and would have learned nothing.

````python
print("Number of features:", X.shape[1])
print()
print("Feature names:")
for name in X.columns:
    print(" -", name)
````

*(ANSWER — expected output)*

**9 features:** `bill_length_mm`, `bill_depth_mm`, `flipper_length_mm`, `body_mass_g`, `island_Biscoe`, `island_Dream`, `island_Torgersen`, `sex_FEMALE`, `sex_MALE`.

We started with 6 columns. The 4 numeric ones passed through untouched; `island` became 3 columns and `sex` became 2, so 2 text columns turned into 5 numeric ones.

### 9.4 Split the Data into Training and Testing Sets

We hold back part of the data so we can test the model on penguins it has never seen.

- `test_size=0.25` holds back 25% of the rows for testing.
- `random_state=42` makes the split reproducible, so everyone in class gets the same result.
- `stratify=y` keeps the mix of species roughly the same in both halves.

````python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

print("Training rows:", len(X_train))
print("Testing rows: ", len(X_test))
print()
print("Species in the training set:")
print(y_train.value_counts())
print()
print("Species in the testing set:")
print(y_test.value_counts())
````

*(ANSWER — expected output)*

**249 training rows** and **84 testing rows**, which together account for all 333 rows.

| Species | Training | Testing |
|---|---|---|
| Adelie | 109 | 37 |
| Gentoo | 89 | 30 |
| Chinstrap | 51 | 17 |

That is `stratify=y` doing its job. Each species holds roughly the same share in both halves — Adelie is about 44% of the training rows and about 44% of the testing rows. Without stratification, a random split could easily have handed us a test set with very few Chinstraps, and our scores would then depend more on luck than on the model.

### 9.5 Train the Model

We will use a **Random Forest**, which builds a large collection of decision trees and lets them vote on the answer.

`fit()` is the step where the model actually learns. It only ever sees the training data.

````python
model = RandomForestClassifier(n_estimators=200, random_state=42)

model.fit(X_train, y_train)

print("Model trained!")
print("Species this model can predict:", model.classes_)
````

### 9.6 Make Predictions

`predict()` returns the single species the model thinks is most likely.

`predict_proba()` returns the model's confidence in *each* species. We need those probabilities later to compute AUC.

````python
y_pred = model.predict(X_test)
y_proba = model.predict_proba(X_test)

# Put the real answers and the predictions side by side
comparison = pd.DataFrame({
    "actual": y_test.values,
    "predicted": y_pred
})

comparison.head(10)
````

````python
# ANSWER — instead of eyeballing 10 rows, count the disagreements directly
mistakes = comparison[comparison["actual"] != comparison["predicted"]]

print("Test penguins:", len(comparison))
print("Mistakes:", len(mistakes))
mistakes
````

*(ANSWER)*

**Zero mistakes.** Every one of the 84 test penguins was classified correctly, so `mistakes` comes back as an empty DataFrame. Hold that thought until YOUR TURN #5.

### 9.7 Evaluate the Model

Now we score the predictions against the real answers. Each metric answers a slightly different question:

- **Accuracy** — of all the test penguins, what fraction did we get right?
- **Precision** — when the model said "Gentoo," how often was it actually a Gentoo?
- **Recall** — of all the real Gentoos, how many did the model find?
- **F1 score** — a single number that balances precision and recall.
- **AUC** — how well the model separates one species from the others across all confidence levels. 1.0 is perfect, 0.5 is a coin flip.

Because we have three species instead of two, we ask for the `macro` average, which computes the metric for each species and then averages them, treating all three species as equally important.

AUC needs one extra argument for the same reason. With only two possible answers, AUC can read a single column of probabilities. With three, we pass the whole `y_proba` array and set `multi_class="ovr"` — "one versus rest" — which scores each species against the other two and then averages the results.

````python
print("Accuracy :", round(accuracy_score(y_test, y_pred), 3))
print("Precision:", round(precision_score(y_test, y_pred, average="macro"), 3))
print("Recall   :", round(recall_score(y_test, y_pred, average="macro"), 3))
print("F1 score :", round(f1_score(y_test, y_pred, average="macro"), 3))
print("AUC      :", round(roc_auc_score(y_test, y_proba, multi_class="ovr", average="macro"), 3))
````

We can also get precision, recall, and F1 broken out **for each species** in one table.

The `support` column tells us how many test penguins of that species there were.

````python
print(classification_report(y_test, y_pred))
````

### 9.8 Confusion Matrix

A **confusion matrix** shows exactly *where* the model was right and where it was wrong.

Read it like this: each **row** is the species a penguin really was, and each **column** is the species the model guessed. Correct predictions land on the diagonal running from the top left to the bottom right. Anything off that diagonal is a mistake.

````python
cm = confusion_matrix(y_test, y_pred, labels=model.classes_)

# As a table
cm_table = pd.DataFrame(cm, index=model.classes_, columns=model.classes_)
print("Rows = actual species, Columns = predicted species")
print(cm_table)
````

````python
display = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=model.classes_
)

display.plot(cmap="Blues")
plt.title("Confusion Matrix: Actual vs. Predicted Species")
plt.show()
````

*(ANSWER — expected output)*

```

Adelie         37          0       0
Chinstrap       0         17       0
Gentoo          0          0      30
```

Every count sits on the diagonal and every off-diagonal cell is 0, which is what a 1.000 accuracy looks like drawn out. The plot shows the same thing: three dark squares running corner to corner, everything else empty.

This is worth seeing precisely because it is boring. In Section 9.11 we will build a model that makes real mistakes, and the contrast between the two matrices is the lesson.

### 9.10 Predict a Brand New Penguin

A model is only useful if we can hand it something new.

Below we invent a penguin that is not in the dataset and ask the model to classify it.

One detail matters here: the new penguin has to be described with **exactly the same columns, in the same order**, that the model was trained on. `reindex()` lines the columns up and fills in any missing category column with 0.

````python
new_penguin = pd.DataFrame([{
    "island": "Biscoe",
    "bill_length_mm": 45.0,
    "bill_depth_mm": 15.0,
    "flipper_length_mm": 220.0,
    "body_mass_g": 5200.0,
    "sex": "MALE"
}])

# Encode it the same way we encoded the training data
new_penguin = pd.get_dummies(new_penguin, columns=["island", "sex"])
new_penguin = new_penguin.reindex(columns=X.columns, fill_value=0)

prediction = model.predict(new_penguin)[0]
probabilities = pd.Series(model.predict_proba(new_penguin)[0], index=model.classes_)

print("Predicted species:", prediction)
print()
print("How confident the model is:")
print(probabilities.round(3))
````

````python
# ANSWER — a helper so we can test several penguins quickly
def classify_penguin(island, bill_length, bill_depth, flipper_length, body_mass, sex):
    row = pd.DataFrame([{
        "island": island,
        "bill_length_mm": bill_length,
        "bill_depth_mm": bill_depth,
        "flipper_length_mm": flipper_length,
        "body_mass_g": body_mass,
        "sex": sex
    }])
    row = pd.get_dummies(row, columns=["island", "sex"])
    row = row.reindex(columns=X.columns, fill_value=0)

    probs = pd.Series(model.predict_proba(row)[0], index=model.classes_)
    print("Predicted:", model.predict(row)[0], " | confidence:", round(probs.max(), 3))
    print(probs.round(3).to_string())
    print()


# A confident penguin: clearly Gentoo
print("--- Clearly Gentoo ---")
classify_penguin("Biscoe", 47.0, 15.0, 218.0, 5000.0, "MALE")

# An ambiguous penguin: sits right on the Adelie / Chinstrap boundary
print("--- Ambiguous ---")
classify_penguin("Dream", 42.5, 16.0, 200.0, 3900.0, "MALE")
````

### 9.11 A Score That Is Too Good

Our first model scored **perfectly**: accuracy 1.000, AUC 1.000, not one mistake in 84 test penguins. That should raise an eyebrow, not a cheer. Real-world data is rarely that tidy, and a perfect score usually means the problem was easy, not that the model was brilliant.

Here the problem really is easy. The three species have noticeably different bills and flippers, so they are simple to tell apart — Section 8 showed us that before we trained anything at all.

To see what a normal score looks like, let's keep the **exact same data, the exact same model, and the exact same code**, and change only one thing: instead of predicting the species, we will predict the penguin's **sex**.

That is a genuinely harder question. Male penguins tend to be a bit larger than females of the same species, but the two overlap a great deal, so there is no clean dividing line for the model to find.

````python
# Same rows, same model settings. Only the target changes.
y_sex = model_df["sex"]

sex_feature_columns = [
    "bill_length_mm",
    "bill_depth_mm",
    "flipper_length_mm",
    "body_mass_g",
    "species",
    "island"
]

X_sex = model_df[sex_feature_columns]
X_sex = pd.get_dummies(X_sex, columns=["species", "island"])

X_sex_train, X_sex_test, y_sex_train, y_sex_test = train_test_split(
    X_sex,
    y_sex,
    test_size=0.25,
    random_state=42,
    stratify=y_sex
)

sex_model = RandomForestClassifier(n_estimators=200, random_state=42)
sex_model.fit(X_sex_train, y_sex_train)

y_sex_pred = sex_model.predict(X_sex_test)
y_sex_proba = sex_model.predict_proba(X_sex_test)

print("What this model can predict:", sex_model.classes_)
print()
print("Accuracy :", round(accuracy_score(y_sex_test, y_sex_pred), 3))
print("Precision:", round(precision_score(y_sex_test, y_sex_pred, pos_label="MALE"), 3))
print("Recall   :", round(recall_score(y_sex_test, y_sex_pred, pos_label="MALE"), 3))
print("F1 score :", round(f1_score(y_sex_test, y_sex_pred, pos_label="MALE"), 3))
print("AUC      :", round(roc_auc_score(y_sex_test, y_sex_proba[:, 1]), 3))
print()
print(classification_report(y_sex_test, y_sex_pred))
````

Notice two small changes in the code above.

Because there are only **two** possible answers now, we no longer need `average="macro"` — instead we tell precision, recall, and F1 which class counts as the positive one with `pos_label="MALE"`. And AUC goes back to reading a single column of probabilities, `y_sex_proba[:, 1]`, with no `multi_class` argument.

This is worth pausing on: the *metric* did not change, only the *shape* of the problem did. Two-class and three-class problems are scored with the same ideas but slightly different arguments.

````python
cm_sex = confusion_matrix(y_sex_test, y_sex_pred, labels=sex_model.classes_)

print("Rows = actual sex, Columns = predicted sex")
print(pd.DataFrame(cm_sex, index=sex_model.classes_, columns=sex_model.classes_))
print()

ConfusionMatrixDisplay(
    confusion_matrix=cm_sex,
    display_labels=sex_model.classes_
).plot(cmap="Oranges")

plt.title("Confusion Matrix: Actual vs. Predicted Sex")
plt.show()
````

````python
# ANSWER — where do the mistakes fall? Break the errors out by species.
sex_results = pd.DataFrame({
    "actual": y_sex_test.values,
    "predicted": y_sex_pred,
    "species": model_df.loc[X_sex_test.index, "species"].values
})
sex_results["correct"] = sex_results["actual"] == sex_results["predicted"]

print(sex_results.groupby("species")["correct"].agg(["sum", "count", "mean"]).round(3))
print()
print("The six penguins the model got wrong:")
print(sex_results[~sex_results["correct"]])
````

Six mistakes out of 84 — four females called male, two males called female. Compare that to Section 9.8, where every off-diagonal cell was 0.

**Discussion questions and answers:**

- **Is 0.929 a bad score?** No — it is a *believable* score. Predicting a penguin's sex from four body measurements and getting it right 93% of the time is a real result. The 1.000 from Section 9 told us almost nothing except that the species were far apart to begin with.

- **Where do the mistakes fall?** Not evenly. Broken out by species, the model gets **Adelie** right 31 of 32 times (96.9%) and **Gentoo** 34 of 36 (94.4%), but **Chinstrap** only 13 of 16 (81.2%). Three of the six errors are Chinstraps, even though Chinstraps are the smallest group in the test set. The single headline accuracy number completely hides that.

- **Which features did the model use?** Run `pd.Series(sex_model.feature_importances_, index=X_sex.columns).sort_values(ascending=False)` and you get `bill_depth_mm` (0.298) and `body_mass_g` (0.298) at the top, with the species and island columns near zero. That matches the biology: within any species, males have deeper bills and more mass. It is also a nice mirror of Section 9.9, where `bill_depth_mm` mattered much less and `sex` mattered almost not at all. **The same column can be crucial for one question and useless for another.** Importance is a property of the question, not of the column.

- **Why does the AUC (0.963) sit higher than the accuracy (0.929)?** Accuracy only asks whether the final answer was right. AUC looks at the underlying probabilities and asks whether the model ranked males above females overall. A model can rank well and still land on the wrong side of the 0.5 cutoff for a handful of borderline penguins — which is exactly what happened to those six.

**The lesson for the whole semester:** we never judge a model by a single number. Accuracy hides *which* cases fail. The per-class report and the confusion matrix are what reveal it, and knowing which cases fail is usually what determines whether a model is safe to actually use.
